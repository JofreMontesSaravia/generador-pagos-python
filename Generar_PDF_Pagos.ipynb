```python
from google.colab import auth, drive
import google.auth
import pandas as pd
import gspread
from PIL import Image, ImageDraw, ImageFont
import os

drive.mount('/content/drive')

os.chdir('ruta de la carpeta del proyecto')

# ruta de carpeta donde se guardaran los pdf
carpeta_doc = "pago-doc"

# crear la carpeta si no existe
if not os.path.exists(carpeta_doc):
    os.mkdir(carpeta_doc)

# recorremos el data frame

plantilla = Image.open("Recibo-Plantilla.png").convert("RGB")

auth.authenticate_user()
creds, _= google.auth.default()
gc = gspread.authorize(creds)

spreadsheet = gc.open("Presupuesto")
worksheet_pago = spreadsheet.worksheet("pago")
data_pago = worksheet_pago.get_all_records()
df_pago = pd.DataFrame(data_pago)

coordenadas = {
    'Fecha': (1020,311),
    'ID': (1280,311),
    'Nombre': (147,530),
    'Teléfono':(495,530),
    'Dirección':(811,530),
    'Horas Trabajadas': (1230,800),
    'Pago por Hora': (1180,885),
    'Pago Total': (1180,1025)
}

fuente = ImageFont.truetype('font.ttf',25)

for _, empleado in df_pago.iterrows():
  imagen = plantilla.copy()
  draw = ImageDraw.Draw(imagen)
  for campo, coordenada in coordenadas.items():
    draw.text(coordenada,
              str(empleado[campo]),
              font= fuente,
              fill="#745151")
  # Nombre del archivo
  nombre = str(empleado["Nombre"]).strip()
  nombre_archivo = f"{nombre}.pdf"

  # Ruta completa
  ruta_pdf = os.path.join(carpeta_doc, nombre_archivo)

  # Guardar PDF
  imagen.save(
  ruta_pdf,
  "PDF",
  resolution=100.0)

print("Todos los PDF fueron generados correctamente.")
```
